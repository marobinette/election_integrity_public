```python
import pandas as pd
df = pd.read_csv('./combineddata.csv')
```

```python
unnamed = df.columns.str.contains('^Unnamed')
df_clean = df.loc[:, ~unnamed].copy()
df_clean.info()
```

```python
pat = r'immigra(?:nt|tion)'

text = (df_clean['Subject'].fillna('') + ' '
        + df_clean['Body'].fillna(''))

mask = text.str.contains(pat, case=False, regex=True, na=False)
df_imm = df_clean[mask].copy()

print(len(df_imm), len(df_imm) / len(df_clean))
```
